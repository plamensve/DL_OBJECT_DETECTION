## Snickers, Mars, Bounty, Twix – Object Detection & Classification
---

### Project Purpose
This project builds an object detection system that recognizes and localizes four chocolate products (Snickers, Mars, Bounty, Twix) in images using deep learning (YOLO). The target scenario is a retail-like environment where items can appear under different lighting conditions and viewing angles.

Key goals:
- detect objects with bounding boxes;
- classify each detected product correctly;
- provide a repeatable workflow for data preparation, training, evaluation, and inference.

### End-to-End Workflow
The project follows a structured and reproducible pipeline:

1. **Data Collection**
   - Gather or download product images.
   - Store raw images in `03_source_data/`.

2. **Labeling (Annotation)**
   - Manually annotate product locations.
   - Save images, labels, and class lists in `04_labeled_data/`.

3. **Dataset Preparation**
   - Use scripts in `00_data_preparation/` to prepare data.
   - Steps include:
     - split into train/validation sets;
     - generate a `data.yaml` configuration for YOLO.

4. **Model Training**
   - Run experiments and training in `01_notebooks/`.
   - Track metrics and artifacts for comparison.

5. **Experiment Tracking**
   - Use MLflow components in `05_experiments/`.
   - Log parameters, metrics, and trained models.

6. **Inference & Visualization**
   - Apply the trained model to new images (in `photos/`).
   - Visualize predictions with bounding boxes and confidence scores.

### Outcome
The final output is a trained object detection model capable of identifying the four product classes in real-world images, along with a clear, reusable workflow for future experiments and extensions.
