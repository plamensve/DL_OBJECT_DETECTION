### SMBT Object Detection – Image Inference


```python
from ultralytics import YOLO
import cv2
```

```python
model = YOLO("runs/detect/smbt_detection_172/weights/best.pt")
```

```python
image_path = "../photos/mtbs.jpg"
```

```python
CONFIDENCE = 0.3
image = cv2.imread(image_path)

results = model(image, conf=CONFIDENCE)
annotated = results[0].plot()

cv2.namedWindow("SMBT Detection - Image", cv2.WINDOW_NORMAL)
cv2.imshow("SMBT Detection - Image", annotated)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

Rotated bounding boxes could improve object localization when objects are seen at an angle, but they add a lot of complexity to the model architecture and evaluation process. This complexity is not necessary for small and compact retail items. For this reason, this project focuses on improving localization quality by working within the limitations of axis-aligned bounding boxes and modifying the bounding box loss function.
