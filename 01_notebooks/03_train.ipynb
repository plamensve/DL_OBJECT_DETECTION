```python
import os

import ultralytics
from ultralytics import YOLO
```

```python
ultralytics.settings["mlflow"] = False
```

### Overview

This script trains a YOLOv8 object detection model using the Ultralytics framework.
It configures the training environment, loads a pretrained YOLOv8 model, and fine-tunes it on a custom dataset defined by a data.yaml configuration file

02_dataset/data.yaml

```python
model = YOLO("yolov8s.pt")
```

```python
model.train(
    data="../02_dataset/data.yaml",
    epochs=150,
    imgsz=640,
    batch=4,
    device=0,
    workers=0,
    name="smbt_detection_7",

    mosaic=1.0,
    mixup=0.1,
    close_mosaic=10,
    fliplr=0.5,
    hsv_h=0.015,
    hsv_s=0.7,
    hsv_v=0.4
)
```
