```python
from ultralytics import YOLO
```

```python
model = YOLO("yolov8s.pt")
```

```python
model.train(
    data="../02_dataset/data.yaml",
    epochs=200,
    imgsz=960,
    batch=4,
    device=0,
    workers=0,
    name="smbt_detection5",

    # augmentation
    mosaic=1.0,
    mixup=0.1,
    fliplr=0.5,
    hsv_h=0.015,
    hsv_s=0.7,
    hsv_v=0.4
)
```
