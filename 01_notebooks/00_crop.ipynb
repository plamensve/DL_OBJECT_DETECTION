```python
import cv2
import numpy as np
from pathlib import Path
```

```python
SRC_DIR = Path(r"/source_data/png")
OUT_DIR = Path(r"/04_labeled_data\images")
OUT_DIR.mkdir(parents=True, exist_ok=True)

EXTS = {".png", ".jpg", ".jpeg", ".bmp", ".webp"}

count = 0

for img_path in SRC_DIR.rglob("*"):
    if img_path.suffix.lower() not in EXTS:
        continue

    print("Processing:", img_path)

    img = cv2.imread(str(img_path), cv2.IMREAD_COLOR)
    if img is None:
        print("  -> FAILED TO READ")
        continue

    h, w = img.shape[:2]

    # ТIGHT CROP – само зоната със символа
    cw = int(w * 0.15)   # ширина
    ch = int(h * 0.28)   # височина

    corner = img[0:ch, 0:cw]

    out_name = img_path.stem + "_corner.jpg"
    cv2.imwrite(str(OUT_DIR / out_name), corner)

    count += 1

print(f"DONE. Generated {count} images.")
```
